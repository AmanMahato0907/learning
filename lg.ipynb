```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# Load dataset
df=pd.read_excel("C:\\Users\\LENOVO\\Downloads\\pima-indians-diabetes.xlsx")

# Select columns
X = df[["age", "Plas"]]   # independent variables
y = df["class"]           # dependent variable

# -------------------------
# BEFORE Logistic Regression
# -------------------------
plt.figure()
plt.scatter(X["age"], X["Plas"])
plt.xlabel("Age")
plt.ylabel("Plasma Glucose")
plt.title("Before Logistic Regression (Raw Data)")
plt.show()

# -------------------------
# Apply Logistic Regression
# -------------------------
model = LogisticRegression()
model.fit(X, y)

# Create grid for decision boundary
x_min, x_max = X["age"].min()-1, X["age"].max()+1
y_min, y_max = X["Plas"].min()-1, X["Plas"].max()+1

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 200),
    np.linspace(y_min, y_max, 200)
)

Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# -------------------------
# AFTER Logistic Regression
# -------------------------
plt.figure()
plt.contour(xx, yy, Z)
plt.scatter(X["age"], X["Plas"])
plt.xlabel("Age")
plt.ylabel("Plasma Glucose")
plt.title("After Logistic Regression (Decision Boundary)")
plt.show()
```
