Q20

```python
a=np.arange(1,10).reshape(3,3)
b=np.arange(10,19).reshape(3,3)
c=np.zeros((3,3),dtype=int)
for i in range(3):
    for j in range(3):
        for k in range(3):
            c[i][j]+=a[i][k]*b[k][j]
print(c)


```
